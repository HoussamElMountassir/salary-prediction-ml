# 📊 Exploratory Data Analysis - Salary Prediction Project

**Objective**: Comprehensive analysis of salary data to understand patterns, relationships, and insights

**Dataset**: 6,704 employee records across 5 countries with demographic and professional factors

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("📚 Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv('../data/raw/Salary_Data_Based_country_and_race.csv')

print("🎯 Dataset Overview")
print(f"Shape: {df.shape}")
print(f"Size: {df.size:,} total values")
print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Display first few rows
print("\n📋 First 5 rows:")
df.head()
```

```python
# Basic dataset information
print("🔍 Dataset Information:")
print(df.info())

print("\n📊 Statistical Summary:")
df.describe(include='all')
```

```python
# Missing values analysis
print("❓ Missing Values Analysis:")
missing_data = df.isnull().sum()
missing_percentage = (missing_data / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_data,
    'Percentage': missing_percentage
}).sort_values('Missing Count', ascending=False)

print(missing_df[missing_df['Missing Count'] > 0])

# Visualize missing values
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=True, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Salary distribution analysis
print("💰 Salary Distribution Analysis:")
salary_stats = df['Salary'].describe()
print(salary_stats)

# Create salary distribution plots
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Histogram
axes[0, 0].hist(df['Salary'], bins=50, alpha=0.7, color='skyblue', edgecolor='black')
axes[0, 0].set_title('Salary Distribution (Histogram)')
axes[0, 0].set_xlabel('Salary ($)')
axes[0, 0].set_ylabel('Frequency')

# Box plot
axes[0, 1].boxplot(df['Salary'], vert=True)
axes[0, 1].set_title('Salary Distribution (Box Plot)')
axes[0, 1].set_ylabel('Salary ($)')

# Log-transformed histogram
axes[1, 0].hist(np.log1p(df['Salary']), bins=50, alpha=0.7, color='lightgreen', edgecolor='black')
axes[1, 0].set_title('Log-Transformed Salary Distribution')
axes[1, 0].set_xlabel('Log(Salary + 1)')
axes[1, 0].set_ylabel('Frequency')

# Q-Q plot
from scipy import stats
stats.probplot(df['Salary'], dist="norm", plot=axes[1, 1])
axes[1, 1].set_title('Q-Q Plot (Normal Distribution)')

plt.tight_layout()
plt.show()
```
